```python
produk_clean = produk[["Uniqe Id", "Product Name", "Category", "Selling Price"]].copy()
```

```python
produk_clean = produk_clean.drop_duplicates(subset="Uniqe Id")
```

```python
produk_clean["Selling Price"] = (
    produk_clean["Selling Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)
produk_clean["Selling Price"] = pd.to_numeric(produk_clean["Selling Price"], errors="coerce")
```

```python
produk_clean["main_category"] = produk_clean["Category"].str.split("|").str[0].str.strip()
```

```python
produk_clean["sub_category"] = produk_clean["Category"].str.split("|").str[1].str.strip()
```

```python
produk_clean["sub_category"] = produk_clean["Category"].str.split("|").str[1].str.strip()
```

```python
produk_clean = produk_clean.rename(columns={
    "Uniqe Id": "product_id",
    "Product Name": "product_name",
    "Selling Price": "price"
})
produk_clean.info()
```

```python
produk_clean["main_category"].value_counts().head(10)
```

```python
customer_clean = customer.copy()
customer_clean.columns = (
    customer_clean.columns.str.strip().str.lower().str.replace(" ", "_", regex=False)
)
customer_clean.columns.tolist()
```

```python
data = interaksi_clean.merge(produk_clean, on="product_id", how="left")
print(data.shape)
```

```python
data["product_name"].notnull().sum()
```

```python
data.isnull().sum()
```

```python
data_f = data[data["interaction_type"].notnull()].copy()
print(data_f.shape)
```

```python
funnel = data_f["interaction_type"].value_counts()
funnel_pct = (funnel / funnel.sum() * 100).round(1)
pd.DataFrame({"jumlah": funnel, "persen": funnel_pct})
```

```python
tabel = pd.crosstab(data_f["sub_category"], data_f["interaction_type"])
tabel["total"] = tabel.sum(axis=1)
tabel["persen_purchase"] = (tabel["purchase"] / tabel["total"] * 100).round(1)
tabel[tabel["total"] >= 20].sort_values("persen_purchase", ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt

funnel.plot(kind="bar", color=["#4C78A8", "#F58518", "#54A24B"])
plt.title("Jumlah Interaksi per Jenis")
plt.xlabel("Jenis interaksi")
plt.ylabel("Jumlah")
plt.xticks(rotation=0)
plt.show()
```

```python
data_f["bulan"] = data_f["timestamp"].dt.to_period("M")
tren = pd.crosstab(data_f["bulan"], data_f["interaction_type"])
tren
```

```python
tren.plot(kind="line", marker="o", figsize=(10, 4))
plt.title("Tren Interaksi per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah")
plt.show()
```
